```python
paths = {
    "synthetic": REPO_ROOT / "results" / "synthetic_optimization" / "aggregate.csv",
    "synthetic_smoke": REPO_ROOT / "results" / "difficult_optimization_smoke" / "aggregate.csv",
    "transformer": REPO_ROOT / "results" / "tinygpt_optimizer_benchmark" / "aggregate.csv",
    "transformer_smoke": REPO_ROOT / "results" / "tinygpt_optimizer_benchmark_smoke" / "aggregate.csv",
}

tables = {}
if pd is None:
    print("Install pandas for combined analysis tables.")
else:
    for name, path in paths.items():
        if path.exists():
            tables[name] = pd.read_csv(path)
            print(f"Loaded {name}: {path}")
```
